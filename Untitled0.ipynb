```python
import pandas as pd
import matplotlib.pyplot as plt

GARIS = "=" * 60

def judul(teks):
    print("\n" + GARIS)
    print(f"  {teks}")
    print(GARIS)

# A. PENGORGANISASIAN DATA

# Tabel Data (file excel satu folder dengan script ini)
path = "data.xlsx"
dataraw = pd.read_excel(path)

judul("A1. TABEL DATA MENTAH")
print(dataraw.to_string(index=False))

# Tabel Frekuensi
datafrq = pd.crosstab(index=dataraw["Grade"], columns="Frekuensi")
datafrq.columns.name = None

# Modifikasi: tambah persentase & total
tabel = datafrq.copy()
tabel["Persentase (%)"] = (tabel["Frekuensi"] / tabel["Frekuensi"].sum() * 100).round(1)
tabel.loc["Total"] = [tabel["Frekuensi"].sum(), 100.0]
tabel["Frekuensi"] = tabel["Frekuensi"].astype(int)

judul("A2. TABEL FREKUENSI GRADE")
print(tabel)

# B. PENYAJIAN DATA (GRAFIK)

warna = ["#4C72B0", "#55A868", "#8172B2", "#64B5CD", "#CCB974", "#DD8452", "#C44E52"]

# Grafik Garis
ax = datafrq.plot(marker="o", color="#4C72B0", linewidth=2,
                  title="Grafik Garis Frekuensi Grade", grid=True)
ax.set_ylabel("Jumlah Mahasiswa")
ax.get_figure().savefig("grafik_garis.png", dpi=150, bbox_inches="tight")

# Grafik Batang
plt.figure()
ax = datafrq["Frekuensi"].plot(kind="bar", color=warna[:len(datafrq)], rot=0,
                  title="Grafik Batang Frekuensi Grade")
ax.set_ylabel("Jumlah Mahasiswa")
for p in ax.patches:  # label angka di atas batang
    ax.annotate(int(p.get_height()), (p.get_x() + p.get_width() / 2, p.get_height()),
                ha="center", va="bottom")
ax.get_figure().savefig("grafik_batang.png", dpi=150, bbox_inches="tight")

# Pie Chart
plt.figure()
ax = datafrq.plot(kind="pie", y="Frekuensi", autopct="%1.0f%%", legend=False,
                  colors=warna[:len(datafrq)], startangle=90,
                  wedgeprops={"edgecolor": "white"}, pctdistance=0.78,
                  title="Pie Chart Persentase Grade")
ax.set_ylabel("")
ax.get_figure().savefig("pie_chart.png", dpi=150, bbox_inches="tight")

# C. STATISTIKA DESKRIPTIF-
dataraw["Nilai"] = pd.to_numeric(dataraw["Nilai"], errors="coerce")
dt = dataraw["Nilai"]
stats = dt.describe()
stats["Standard Error"] = dt.sem()
stats["Median"] = dt.median()
stats["Mode"] = dt.mode().iloc[0]
stats["variance"] = dt.var()
stats["range"] = dt.max() - dt.min()
stats["skewness"] = dt.skew()
stats["kurtosis"] = dt.kurtosis()

judul("C1. STATISTIKA DESKRIPTIF NILAI AKHIR")
print(stats.round(4).to_string())

# Modifikasi: analisis tambahan
q1, q3 = dt.quantile(0.25), dt.quantile(0.75)
iqr = q3 - q1
bawah, atas = q1 - 1.5 * iqr, q3 + 1.5 * iqr
outlier = dataraw[(dt < bawah) | (dt > atas)]
cv = dt.std() / dt.mean() * 100

judul("C2. ANALISIS TAMBAHAN")
print(f"Semua modus          : {', '.join(f'{m:.2f}' for m in dt.mode())}")
print(f"IQR (Q3 - Q1)        : {iqr:.2f}")
print(f"Batas bawah outlier  : {bawah:.2f}")
print(f"Batas atas outlier   : {atas:.2f}")
print(f"Jumlah outlier       : {len(outlier)}")
print(f"Koefisien variasi    : {cv:.2f}%")
print(f"Mean vs Median       : {'Mean < Median -> menceng ke kiri' if dt.mean() < dt.median() else 'Mean >= Median'}")

plt.show()
```
